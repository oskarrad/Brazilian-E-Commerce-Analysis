# Data Analysis in Python

These datasets contain information about orders, customers, sellers, order items, and reviews, which will be used in further analysis.

```python
import pandas as pd
import sqlite3
```

## Extract
In this step, the required datasets are loaded using pandas and raw data is exported to database.


```python
customers = pd.read_csv('../data/olist_customers_dataset.csv')
order_items = pd.read_csv('../data/olist_order_items_dataset.csv')
reviews = pd.read_csv('../data/olist_order_reviews_dataset.csv')
orders = pd.read_csv('../data/olist_orders_dataset.csv')
sellers = pd.read_csv('../data/olist_sellers_dataset.csv')
products = pd.read_csv('../data/olist_products_dataset.csv')
translation = pd.read_csv('../data/product_category_name_translation.csv', header=0)


conn = sqlite3.connect("databaseRawData.db")

customers.to_sql("customers", conn, if_exists="replace", index=False)
order_items.to_sql("order_items", conn, if_exists="replace", index=False)
reviews.to_sql("reviews", conn, if_exists="replace", index=False)
orders.to_sql("orders", conn, if_exists="replace", index=False)
sellers.to_sql("sellers", conn, if_exists="replace", index=False)
products.to_sql("products", conn, if_exists="replace", index=False)
translation.to_sql("translation", conn, if_exists="replace", index=False)
```

## Transformation
After the extraction phase, it’s time to transform the data by identifying relationships between the data points and creating new features that will be useful for future analysis.



The first step will be to gather detailed information about specific datasets and identify the necessary features.

```python
customers.info()
customers = customers.loc[:,["customer_id", "customer_city", "customer_state"]]
customers.info()
```

```python
order_items.info()
order_items = order_items.loc[:,["order_id","product_id","seller_id","price","freight_value"]]
order_items.info()
```

```python
reviews.info()
reviews = reviews.loc[:,["order_id","review_score"]]
reviews.info()
```

```python
orders.info()
orders = orders.loc[:,["order_id","customer_id","order_delivered_carrier_date","order_delivered_customer_date","order_estimated_delivery_date"]]
orders.info()
```

```python
sellers.info()
sellers = sellers.loc[:,["seller_id","seller_city","seller_state"]]
sellers.info()
```

```python
products.info()
products = products.loc[:,["product_id","product_category_name"]]
products.info()
```

```python
translation.info()
```

After extracting the necessary attributes, we then join the datasets using primary and foreign keys.

```python
#Left Outer Join orders and order_items ON order_id
result = pd.merge(orders, order_items, on="order_id", how="left")

#Left Outer Join result and reviews ON order_id
result = pd.merge(result, reviews, on="order_id", how="left")

#Left Outer Join result and customers ON customer_id
result = pd.merge(result, customers, on="customer_id", how="left")

#Left Outer Join result and sellers ON seller_id
result = pd.merge(result, sellers, on="seller_id", how="left")

#Left Outer Join result and products ON product_id
result = pd.merge(result, products, on="product_id", how="left")

#Left Outer Join result and product_category ON product_category_name
result = pd.merge(result, translation, on="product_category_name", how="left")

result["total_order_value"] = result["price"] + result["freight_value"]
result.rename(columns={"product_category_name_english": "category"}, inplace=True)
result.drop("product_category_name", axis=1, inplace=True)
date_cols = [
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_cols:
    result[col] = pd.to_datetime(result[col])
```

After the merge, the final table was reviewed, and the number of NULL values for each column was checked

```python
result.info()
result.isnull().sum()
```

As you can see, there are quite a few NULL values, particularly in the 'order_delivered_customer_date' column, which contains 3,253 such values.

## Handling Missing Values

Missing values were analyzed and handled based on their meaning:

- Orders without delivery dates were removed, as they were not relevant for delivery analysis
- Rows without price or freight value were removed, as they cannot be used for revenue calculations
- Missing review scores were kept, as not all customers leave reviews

```python
result = result.dropna(subset=["price", "freight_value"])
result = result.dropna(subset=["order_delivered_customer_date", "order_delivered_carrier_date"])

result.info()
```

Next, new features were created that will be necessary for future analysis.

```python
#delivery_delay_days
result["delivery_delay_days"] = (result["order_delivered_customer_date"] - result["order_estimated_delivery_date"]).dt.days
result.loc[result["delivery_delay_days"] < 0, "delivery_delay_days"] = 0

#on_time
result["on_time"] = (result["order_estimated_delivery_date"] >= result["order_delivered_customer_date"]).astype(int)

#same_state
result["same_state"] = (result["seller_state"] == result["customer_state"]).astype(int)

#delivery_time
result["delivery_time"] = (result["order_delivered_customer_date"] - result["order_delivered_carrier_date"]).dt.days

result.info()
```

The previously created table contains multiple rows for a single order, breaking them down by individual products. Therefore, a separate table for grouped orders has been created. It contains the following columns: 'order_id', 'total_order_value', 'total_freight_price', 'items_count'.

```python
result_grouped = result.groupby("order_id")[["order_id","price","freight_value"]].agg({
    "order_id":"count",
    "price":"sum",
    "freight_value":"sum",
    })
result_grouped.rename(columns={"order_id":"items_count"}, inplace=True)
result_grouped["total_order_value"] = (result_grouped["price"] + result_grouped["freight_value"])
result_grouped.reset_index(inplace=True)

result_grouped.head()
```

## Load
Data was loaded into a SQLite for further analysis.

```python
conn = sqlite3.connect("databaseTransformedData.db")

result_grouped.to_sql("orders_grouped", conn, if_exists="replace", index=False)
result.to_sql("orders", conn, if_exists="replace", index=False)
```

## Analysis
The next step will be to conduct an analysis in an effort to find answers to the questions raised.

### Which product categories generate the highest revenue?

```python
revenue = (pd.pivot_table(result, values="total_order_value", index="category", aggfunc="sum")
           .sort_values(by=["total_order_value"], ascending=False))
print(revenue.head(15))
```

The highest revenue is generated by categories such as **health_beauty**, **watches_gifts**, and **bed_bath_table**.
These categories dominate the e-commerce market, indicating strong demand for lifestyle and personal-use products.

### Which locations have the highest sales?


```python
sales = pd.pivot_table(result, values="total_order_value", index=["seller_state","seller_city"], aggfunc="sum")
state_totals = sales.groupby(["seller_state"])["total_order_value"].sum()
sales["state_total"] = sales.index.get_level_values("seller_state").map(state_totals)
sales_sorted = sales.sort_values(by=["state_total","total_order_value"], ascending=[False,False])
sales_sorted.drop(columns=["state_total"], inplace=True)
print(sales_sorted)
```

Sales are highly concentrated in a few regions, with **São Paulo (SP)** leading by a significant margin.
Other major contributors include **Paraná (PR)** and **Minas Gerais (MG)**.

This indicates a strong regional concentration of e-commerce activity, with large urban areas dominating total sales.

### What is the average delivery time by Location?

```python
delivery = pd.pivot_table(result, values="delivery_time", index="same_state", aggfunc="mean").round(2)
print(delivery)
```

Average delivery time differs significantly depending on location:

- Same state deliveries are considerably faster
- Cross-state deliveries take more than twice as long on average

This highlights the strong impact of geographical distance on logistics performance.

### Are there delays in delivery by Location?

```python
delays = pd.pivot_table(result, values=["delivery_delay_days","review_score"], index="same_state", aggfunc="mean").round(2)
print(delays)
```

Delivery delays are generally low, but are noticeably higher for cross-state orders.

Additionally, customer satisfaction (review score) is slightly lower when delays occur, suggesting that delivery performance has a direct impact on customer experience.

## Conclusion

The analysis shows that delivery performance and sales are strongly influenced by geographical factors.

- Sales are concentrated in major regions such as São Paulo
- Cross-state deliveries take significantly longer and experience higher delays
- Faster deliveries are associated with higher customer satisfaction

These findings highlight the importance of logistics optimization in improving both operational efficiency and customer experience.
